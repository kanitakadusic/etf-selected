```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
```

# Zadatak 1

```python
hrana = ctrl.Antecedent(np.arange(0, 11, 1), "hrana")
usluga = ctrl.Antecedent(np.arange(0, 11, 1), "usluga")
napojnica = ctrl.Consequent(np.arange(0, 26, 1), "napojnica")
```

```python
hrana.automf(3)
usluga.automf(3)

napojnica["mala"] = fuzz.trimf(napojnica.universe, [0, 0, 13])
napojnica["srednja"] = fuzz.trimf(napojnica.universe, [0, 13, 25])
napojnica["velika"] = fuzz.trimf(napojnica.universe, [13, 25, 25])
```

```python
p1 = ctrl.Rule(usluga["poor"] & hrana["poor"], napojnica["mala"])
p2 = ctrl.Rule(usluga["average"], napojnica["srednja"])
p3 = ctrl.Rule(usluga["good"] | hrana["good"], napojnica["velika"])
```

```python
sistem = ctrl.ControlSystem([p1, p2, p3])
```

```python
simulacija = ctrl.ControlSystemSimulation(sistem)
```

```python
simulacija.input["hrana"] = 6.5
simulacija.input["usluga"] = 9.8

simulacija.compute()
print(simulacija.output["napojnica"])
```

# Zadatak 2

```python
hrana = ctrl.Antecedent(np.arange(0, 11, 1), "hrana")
usluga = ctrl.Antecedent(np.arange(0, 11, 1), "usluga")
napojnica = ctrl.Consequent(np.arange(0, 51, 1), "napojnica")
```

```python
hrana["losa"] = fuzz.trimf(hrana.universe, [0, 0, 4])
hrana["prihvatljiva"] = fuzz.trapmf(hrana.universe, [3, 4, 6, 7])
hrana["odlicna"] = fuzz.trimf(hrana.universe, [6, 10, 10])

usluga["losa"] = fuzz.trimf(usluga.universe, [0, 0, 4])
usluga["prihvatljiva"] = fuzz.trapmf(usluga.universe, [3, 4, 6, 7])
usluga["odlicna"] = fuzz.trimf(usluga.universe, [6, 10, 10])

napojnica["mala"] = fuzz.trimf(napojnica.universe, [0, 0, 25])
napojnica["srednja"] = fuzz.trimf(napojnica.universe, [0, 25, 50])
napojnica["velika"] = fuzz.trimf(napojnica.universe, [25, 50, 50])
```

```python
p1 = ctrl.Rule(usluga["losa"] & hrana["losa"], napojnica["mala"])
p2 = ctrl.Rule(usluga["prihvatljiva"], napojnica["srednja"])
p3 = ctrl.Rule(usluga["odlicna"] | hrana["odlicna"], napojnica["velika"])
```

```python
sistem = ctrl.ControlSystem([p1, p2, p3])
```

```python
simulacija = ctrl.ControlSystemSimulation(sistem)
```

```python
simulacija.input["hrana"] = 6.5
simulacija.input["usluga"] = 9.8

simulacija.compute()
print(simulacija.output["napojnica"])
```
